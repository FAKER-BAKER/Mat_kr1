# Практическое занятие №2: Комплексные числа. Геометрия комплексных чисел

## 1. Алгебраическая, тригонометрическая и показательная формы
* **Алгебраическая форма**: $z = x + iy$, где $x = \mathrm{Re}\,z$, $y = \mathrm{Im}\,z$.
* **Модуль и аргумент**: $|z| = \sqrt{x^2+y^2}$; $\arg z = \varphi$ — угол между радиус-вектором точки $z$ и положительным направлением оси $Ox$ (главное значение $\mathrm{Arg}\,z \in (-\pi, \pi]$).
* **Тригонометрическая форма**: $z = |z|(\cos\varphi + i\sin\varphi)$.
* **Показательная форма**: $z = |z|\,e^{i\varphi}$ (формула Эйлера $e^{i\varphi} = \cos\varphi + i\sin\varphi$).

---

## 2. Операции над комплексными числами
* **Умножение/деление**: $z_1 z_2 = |z_1||z_2|\,e^{i(\varphi_1+\varphi_2)}$, $\dfrac{z_1}{z_2} = \dfrac{|z_1|}{|z_2|}\,e^{i(\varphi_1-\varphi_2)}$.
* **Формула Муавра**: $z^n = |z|^n(\cos n\varphi + i\sin n\varphi) = |z|^n e^{in\varphi}$.
* **Корни $n$-й степени**: уравнение $w^n = z$ имеет ровно $n$ различных решений
  $$w_k = |z|^{1/n}\left(\cos\frac{\varphi+2\pi k}{n} + i\sin\frac{\varphi+2\pi k}{n}\right), \quad k = 0, 1, \dots, n-1,$$
  которые расположены в вершинах правильного $n$-угольника, вписанного в окружность радиуса $|z|^{1/n}$ с центром в начале координат.

---

## 3. Геометрия комплексных чисел
* $|z-a|$ — расстояние от точки $z$ до точки $a$ на плоскости.
* $|z-a| = R$ — окружность радиуса $R$ с центром $a$; $|z-a| < R$ — открытый круг.
* $|z-a| + |z-b| = 2c$ (при $2c > |a-b|$) — эллипс с фокусами $a, b$.
* $\arg\dfrac{z-a}{z-b} = \mathrm{const}$ — дуга окружности, проходящей через точки $a$ и $b$ (геометрическое место точек, из которых отрезок $[a,b]$ виден под фиксированным ориентированным углом); двойное неравенство задаёт "линзу", ограниченную двумя такими дугами.
* Общее уравнение окружности: $A|z|^2 + \overline{B}z + B\overline{z} + C = 0$, где $A, C \in \mathbb{R}$, $B \in \mathbb{C}$. При $A>0$ и $AC < |B|^2$ это уравнение окружности с центром $z_0 = -B/A$ и радиусом $R = \sqrt{|B|^2-AC}/A$.

---

## 4. Многочлены и их корни в $\mathbb{C}$
* **Основная теорема алгебры**: многочлен степени $n$ с комплексными коэффициентами имеет ровно $n$ корней в $\mathbb{C}$ (с учётом кратности).
* Для многочлена с вещественными коэффициентами невещественные корни входят комплексно-сопряжёнными парами.

---

## 5. Гамма-функция и обобщённые биномиальные коэффициенты
* $\Gamma(z) = \displaystyle\int_0^{\infty} x^{z-1}e^{-x}\,dx$, $\mathrm{Re}(z) > 0$; при этом $n! = \Gamma(n+1)$.
* Обобщённый биномиальный коэффициент для $z, w \in \mathbb{C}$: $\displaystyle C_z^w = \frac{\Gamma(z+1)}{\Gamma(w+1)\Gamma(z-w+1)}$.

---

## 6. Сфера Римана и стереографическая проекция
* Сфера Римана — единичная сфера в $\mathbb{R}^3$, точки которой находятся во взаимно однозначном соответствии с точками расширенной комплексной плоскости $\mathbb{C}\cup\{\infty\}$ (точка $\infty$ соответствует северному полюсу).
* Стереографическая проекция точки $z = x+iy$ из северного полюса $N=(0,0,1)$ на сферу:
  $$X = \frac{2x}{1+|z|^2}, \qquad Y = \frac{2y}{1+|z|^2}, \qquad Z = \frac{|z|^2-1}{1+|z|^2}.$$
* При этом прямые и окружности плоскости переходят в окружности на сфере (прямая — в окружность, проходящую через северный полюс $N$).

**2.1** Условие: Представьте в алгебраической, тригонометрической и показательной формах следующие комплексные выражения:
$$a)\ \frac{(5+i)(7-6i)}{3+i}; \qquad b)\ \frac{(1+i)^5}{(1-i)^3}; \qquad c)\ \frac{(-1+i\sqrt3)^{15}}{(1-i)^{20}} + \frac{(-1-i\sqrt3)^{15}}{(1+i)^{20}}.$$

```python
from sympy import I, sqrt, simplify, Abs, arg, N, pi

def show_forms(z, label):
    z = simplify(z)
    r = simplify(Abs(z))
    phi = simplify(arg(z))
    print(f"{label} Алгебраическая форма:      z = {z}")
    print(f"{label} Модуль:                    |z| = {r}  (\u2248 {float(N(r)):.6f})")
    print(f"{label} Аргумент:                  arg z = {phi}  (\u2248 {float(N(phi)):.6f} рад)")
    print(f"{label} Тригонометрическая форма:  z = {r}*(cos({phi}) + i*sin({phi}))")
    print(f"{label} Показательная форма:       z = {r}*exp(i*({phi}))")
    print()

za = (5 + I) * (7 - 6*I) / (3 + I)
show_forms(za, "2.1 a)")

zb = (1 + I)**5 / (1 - I)**3
show_forms(zb, "2.1 b)")

zc = (-1 + I*sqrt(3))**15 / (1 - I)**20 + (-1 - I*sqrt(3))**15 / (1 + I)**20
show_forms(zc, "2.1 c)")
```

**2.2** Условие: Решите уравнение и изобразите множество решений на комплексной плоскости
$$z^8 = 1+i.$$

```python
from sympy import I, Abs, arg, pi, exp, Rational, N, simplify
import numpy as np
import matplotlib.pyplot as plt

w = 1 + I
r = Abs(w)
phi = arg(w)
n = 8

roots = [simplify(r**Rational(1, n) * exp(I*(phi + 2*k*pi)/n)) for k in range(n)]

print(f"2.2 |1+i| = {r} \u2248 {float(N(r)):.6f},   arg(1+i) = {phi}")
print(f"2.2 Модуль каждого корня |z_k| = (1+i)^(1/8) \u2248 {float(N(r**Rational(1, n))):.6f}")
for k, zk in enumerate(roots):
    print(f"    z_{k} \u2248 {complex(N(zk, 8)):.6f}")

roots_num = [complex(N(z)) for z in roots]
xs = [z.real for z in roots_num]
ys = [z.imag for z in roots_num]

fig, ax = plt.subplots(figsize=(6, 6))
theta = np.linspace(0, 2*np.pi, 400)
R = float(N(r**Rational(1, n)))
ax.plot(R*np.cos(theta), R*np.sin(theta), '--', color='gray', linewidth=0.8)
ax.scatter(xs, ys, color='crimson', zorder=3)
for k, (x, y) in enumerate(zip(xs, ys)):
    ax.annotate(f"$z_{{{k}}}$", (x, y), textcoords="offset points", xytext=(6, 6))
ax.axhline(0, color='black', linewidth=0.6)
ax.axvline(0, color='black', linewidth=0.6)
ax.set_aspect('equal')
ax.set_title("2.2 Корни уравнения $z^8=1+i$")
ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
ax.grid(alpha=0.3)
plt.show()
```

**2.3** Условие: Вычислите (над полем $\mathbb{C}$) и изобразите на комплексной плоскости:
$$a)\ \sqrt[8]{1}; \qquad b)\ \sqrt[4]{-\frac{18}{1+i\sqrt3}}; \qquad c)\ \sqrt[3]{\frac{1-5i}{1+i} - 5\frac{1+2i}{2-i} + 11}; \qquad d)\ \sqrt[9]{1-i}.$$

```python
import cmath
import numpy as np
import matplotlib.pyplot as plt

def nth_roots(w, n):
    r = abs(w)
    phi = cmath.phase(w)
    return [r**(1/n) * cmath.exp(1j*(phi + 2*np.pi*k)/n) for k in range(n)]

wa, na = 1, 8
wb, nb = -18/(1 + 1j*np.sqrt(3)), 4
wc, nc = (1-5j)/(1+1j) - 5*(1+2j)/(2-1j) + 11, 3
wd, nd = 1 - 1j, 9

tasks = [("a", wa, na), ("b", wb, nb), ("c", wc, nc), ("d", wd, nd)]

fig, axes = plt.subplots(2, 2, figsize=(11, 11))
for ax, (label, w, n) in zip(axes.ravel(), tasks):
    roots = nth_roots(w, n)
    print(f"2.3 {label}) подкоренное выражение = {w:.6f},  корней: {n}")
    for k, zk in enumerate(roots):
        print(f"    z_{k} \u2248 {zk:.6f}")
    print()

    xs = [z.real for z in roots]
    ys = [z.imag for z in roots]
    R = abs(w)**(1/n)
    theta = np.linspace(0, 2*np.pi, 300)
    ax.plot(R*np.cos(theta), R*np.sin(theta), '--', color='gray', linewidth=0.8)
    ax.scatter(xs, ys, color='crimson', zorder=3)
    for k, (x, y) in enumerate(zip(xs, ys)):
        ax.annotate(f"$z_{{{k}}}$", (x, y), textcoords="offset points", xytext=(5, 5), fontsize=8)
    ax.axhline(0, color='black', linewidth=0.6)
    ax.axvline(0, color='black', linewidth=0.6)
    ax.set_aspect('equal')
    ax.set_title(f"2.3 {label})  $n={n}$")
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

**2.4** Условие: Изобразите на комплексной плоскости множество точек, удовлетворяющих условию:
$$a)\ |z-i|+|z+i|<4; \qquad b)\ \mathrm{Re}(z(1-i))<\sqrt2; \qquad c)\ |z+1|=|z+i|=|z-i|;$$
$$d)\ 0<\arg\frac{i-z}{z+i}<\frac\pi2; \qquad e)\ -\frac{2\pi}3\le\arg\frac{2-iz}{z+3i}\le-\frac{5\pi}6; \qquad f)\ -\frac{2\pi}3\le\arg\frac{3-z}{z+i}\le-\frac\pi3.$$

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_region(cond, xlim, ylim, title, n=700, points=None):
    x = np.linspace(*xlim, n)
    y = np.linspace(*ylim, n)
    X, Y = np.meshgrid(x, y)
    Z = X + 1j*Y
    with np.errstate(all='ignore'):
        mask = cond(Z)
    mask = np.asarray(mask, dtype=float)
    mask[~np.isfinite(mask)] = 0
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.contourf(X, Y, mask, levels=[0.5, 1.5], colors=['#4C72B0'], alpha=0.75)
    ax.axhline(0, color='black', lw=0.6)
    ax.axvline(0, color='black', lw=0.6)
    if points:
        for (px, py), lbl in points:
            ax.plot(px, py, 'ko', ms=4)
            ax.annotate(lbl, (px, py), textcoords="offset points", xytext=(6, 4))
    ax.set_aspect('equal')
    ax.set_title(title, fontsize=10)
    ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
    ax.grid(alpha=0.3)
    plt.show()

# a) |z-i|+|z+i|<4 -- внутренность эллипса с фокусами +-i
plot_region(lambda Z: np.abs(Z - 1j) + np.abs(Z + 1j) < 4, (-4, 4), (-4, 4),
            "2.4 a) $|z-i|+|z+i|<4$", points=[((0, 1), '$i$'), ((0, -1), '$-i$')])

# b) Re(z(1-i)) < sqrt(2) -- полуплоскость
plot_region(lambda Z: (Z*(1 - 1j)).real < np.sqrt(2), (-4, 4), (-4, 4),
            "2.4 b) $\\mathrm{Re}(z(1-i))<\\sqrt{2}$")

# c) |z+1|=|z+i|=|z-i| -- аналитическое решение (единственная точка)
from sympy import symbols, Eq, solve

xs_, ys_ = symbols('x y', real=True)
eq1 = Eq((xs_ + 1)**2 + ys_**2, xs_**2 + (ys_ + 1)**2)
eq2 = Eq(xs_**2 + (ys_ + 1)**2, xs_**2 + (ys_ - 1)**2)
sol_c = solve([eq1, eq2], (xs_, ys_))
z_c = complex(sol_c[xs_], sol_c[ys_])
print(f"2.4 c) |z+1|=|z+i|=|z-i| выполняется только в точке z = {z_c}")

fig, ax = plt.subplots(figsize=(5, 5))
for (px, py), lbl in [((-1, 0), '$-1$'), ((0, -1), '$-i$'), ((0, 1), '$i$')]:
    ax.plot(px, py, 'ko', ms=4)
    ax.annotate(lbl, (px, py), textcoords="offset points", xytext=(6, 4))
ax.plot(z_c.real, z_c.imag, 'r*', ms=16, label=f'$z={z_c.real:.0f}$ \u2014 искомая точка')
theta = np.linspace(0, 2*np.pi, 300)
ax.plot(np.cos(theta), np.sin(theta), '--', color='gray', lw=0.8, label='$|z|=1$')
ax.axhline(0, color='black', lw=0.6)
ax.axvline(0, color='black', lw=0.6)
ax.set_aspect('equal')
ax.set_title("2.4 c) Точка, равноудалённая от $-1, i, -i$")
ax.legend(fontsize=8)
ax.grid(alpha=0.3)
plt.show()

# d) 0 < arg((i-z)/(z+i)) < pi/2
plot_region(lambda Z: (np.angle((1j - Z)/(Z + 1j)) > 0) & (np.angle((1j - Z)/(Z + 1j)) < np.pi/2),
            (-4, 4), (-4, 4), "2.4 d) $0<\\arg\\frac{i-z}{z+i}<\\frac{\\pi}{2}$",
            points=[((0, -1), '$-i$'), ((0, 1), '$i$')])

# e) -2pi/3 <= arg((2-iz)/(z+3i)) <= -5pi/6  (интервал между значениями)
lo_e, hi_e = sorted([-2*np.pi/3, -5*np.pi/6])
plot_region(lambda Z: (np.angle((2 - 1j*Z)/(Z + 3j)) >= lo_e) & (np.angle((2 - 1j*Z)/(Z + 3j)) <= hi_e),
            (-6, 6), (-6, 6), "2.4 e) $-\\frac{2\\pi}{3}\\leq\\arg\\frac{2-iz}{z+3i}\\leq-\\frac{5\\pi}{6}$",
            points=[((0, -3), '$-3i$')])

# f) -2pi/3 <= arg((3-z)/(z+i)) <= -pi/3
lo_f, hi_f = sorted([-2*np.pi/3, -np.pi/3])
plot_region(lambda Z: (np.angle((3 - Z)/(Z + 1j)) >= lo_f) & (np.angle((3 - Z)/(Z + 1j)) <= hi_f),
            (-6, 6), (-6, 6), "2.4 f) $-\\frac{2\\pi}{3}\\leq\\arg\\frac{3-z}{z+i}\\leq-\\frac{\\pi}{3}$",
            points=[((3, 0), '$3$'), ((0, -1), '$-i$')])
```

**2.5\*** Условие: Пусть $A$ и $C$ — действительные, а $B$ — комплексная постоянные, причём $AC < |B|^2$. Покажите, что уравнение
$$A|z|^2 + \overline{B}z + B\overline{z} + C = 0, \qquad (A>0),$$
является уравнением окружности, а также найдите центр этой окружности и её радиус.

*Решение*: делим уравнение на $A$ и дополняем до полного квадрата, используя $|z|^2 = z\overline{z}$:
$$z\overline{z} + \frac{\overline{B}}{A}z + \frac{B}{A}\overline{z} + \frac{C}{A} = \left(z+\frac{B}{A}\right)\overline{\left(z+\frac{B}{A}\right)} - \frac{|B|^2}{A^2} + \frac{C}{A} = 0,$$
то есть
$$\left|z + \frac{B}{A}\right|^2 = \frac{|B|^2 - AC}{A^2}.$$
Поскольку $A>0$ и $AC<|B|^2$, правая часть строго положительна — это уравнение окружности с центром $z_0 = -\dfrac{B}{A}$ и радиусом $R = \dfrac{\sqrt{|B|^2-AC}}{A}$.

```python
import numpy as np

# Численная проверка формулы на конкретном примере
A_val, B_val, C_val = 2.0, 1.5 - 2.0j, 0.5
assert A_val*C_val < abs(B_val)**2, "должно выполняться AC < |B|^2"

z0 = -B_val / A_val
R = np.sqrt(abs(B_val)**2 - A_val*C_val) / A_val

print(f"2.5 Пример: A={A_val}, B={B_val}, C={C_val}")
print(f"2.5 Центр окружности z0 = -B/A = {z0}")
print(f"2.5 Радиус окружности  R = sqrt(|B|^2-AC)/A = {R:.6f}")
print()
print("Проверка: подставляем точки окружности z = z0 + R*e^{i*theta} в исходное уравнение")
for theta in np.linspace(0, 2*np.pi, 6, endpoint=False):
    z = z0 + R*np.exp(1j*theta)
    val = A_val*abs(z)**2 + np.conj(B_val)*z + B_val*np.conj(z) + C_val
    print(f"  theta={theta:.3f}:  A|z|^2+B\u0304z+Bz\u0304+C = {val:.2e}  (\u2248 0 \u2014 верно)")
```

**2.7** (С использованием «Jupyter Notebook») Обобщение биномиальных коэффициентов. Напомним, $C_n^k = \dfrac{n!}{k!(n-k)!}$. Вспомним, что $n!=\Gamma(n+1)$, где
$$\Gamma(z) = \int_0^\infty x^{z-1}e^{-x}\,dx, \qquad \mathrm{Re}(z)>0.$$
Определим формально символ $C_z^w$ для $z,w\in\mathbb{C}$:
$$C_z^w = \frac{\Gamma(z+1)}{\Gamma(w+1)\Gamma(z-w+1)}.$$
a) Найдите $C_{1+i}^{1-i}$.  b) Найдите $\mathrm{Arg}(C_{1+i}^{1-i})$.  c) Постройте линии уровня $\mathrm{Arg}[C_z^{1/z}=\mathrm{const}]$, где $z=x+iy$.

```python
import numpy as np
from scipy.special import loggamma
import matplotlib.pyplot as plt

def C(z, w):
    return np.exp(loggamma(z + 1) - loggamma(w + 1) - loggamma(z - w + 1))

# a), b)
z0, w0 = 1 + 1j, 1 - 1j
val = C(z0, w0)
print(f"2.7 a) C_(1+i)^(1-i) = {val:.5f}")
print(f"2.7 b) Arg(C_(1+i)^(1-i)) = {np.angle(val):.6f} рад ({np.degrees(np.angle(val)):.4f}\u00b0)")

# c) линии уровня Arg[C_z^{1/z}]
nx, ny = 500, 500
x = np.linspace(-6, 6, nx)
y = np.linspace(-6, 6, ny)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y
Z[Z == 0] = np.nan

with np.errstate(all='ignore'):
    W = C(Z, 1/Z)
    Arg = np.angle(W)

fig, ax = plt.subplots(figsize=(8, 8))
im = ax.imshow(Arg, extent=[-6, 6, -6, 6], origin='lower', cmap='twilight')
ax.contour(X, Y, Arg, levels=24, colors='k', linewidths=0.4, alpha=0.6)
ax.set_title("2.7 c) Линии уровня $\\mathrm{Arg}[C_z^{1/z}]$")
ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
plt.colorbar(im, ax=ax, label='Arg')
plt.show()
```

**2.8** Условие: Пусть $\lambda$ — комплексный корень из $1$ степени $14$, такой, что его мнимая часть больше $0$, а действительная часть максимальна среди подобных корней с положительной мнимой частью. Даны матрицы:
$$I=\begin{pmatrix}1&0\\0&1\end{pmatrix}, \quad A=\begin{pmatrix}7&5\\6&2\end{pmatrix}.$$
Найдите определитель матрицы $A+\lambda I$ и укажите в ответе: 1) действительную часть этого определителя; 2) его мнимую часть.

```python
import numpy as np

# Корни из 1 степени 14: exp(i*2*pi*k/14), k=0,...,13.
# Im>0 при k=1,...,6; среди них Re максимальна при наименьшем угле, т.е. k=1.
n = 14
roots14 = [np.exp(1j*2*np.pi*k/n) for k in range(n)]
candidates = [z for z in roots14 if z.imag > 0]
lam = max(candidates, key=lambda z: z.real)

print(f"2.8 Корни степени 14 с Im>0: {[f'{z:.4f}' for z in candidates]}")
print(f"2.8 lambda = {lam:.6f}  (= exp(i*pi/7))")

I2 = np.eye(2)
Amat = np.array([[7, 5], [6, 2]])
M = Amat + lam*I2
det_M = np.linalg.det(M)

print(f"2.8 det(A+\u03bbI) = {det_M:.6f}")
print(f"1) Re(det) = {det_M.real:.6f}")
print(f"2) Im(det) = {det_M.imag:.6f}")
```

г) Напишите программу на Python генерирования множеств Жюлиа для $f(z)=\dfrac{iz^5-1}{z^5-i}$ и $f(z)=\dfrac{z^2+i}{1-z^2}$.

```python
import numpy as np
import matplotlib.pyplot as plt

def julia_set(f, xmin=-2.1, xmax=2.1, ymin=-2.1, ymax=2.1,
              width=400, height=400, max_iter=250, escape_radius=10.0):
    x = np.linspace(xmin, xmax, width)
    y = np.linspace(ymin, ymax, height)
    X, Y = np.meshgrid(x, y)
    Z = X + 1j*Y
    img = np.full(Z.shape, max_iter, dtype=float)
    active = np.ones(Z.shape, dtype=bool)
    with np.errstate(all='ignore'):
        for it in range(max_iter):
            Z[active] = f(Z[active])
            diverged = active & (~np.isfinite(Z) | (np.abs(Z) > escape_radius))
            img[diverged] = it
            active &= ~diverged
    return img / max_iter

def plot_julia(img, win, title):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(img, cmap='flag', extent=[-win, win, -win, win], origin='lower')
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title(title, fontsize=10)
    plt.show()

# в) f_c(z) = z^2 + c
for c in [-0.1 + 0.65j, 0.28 + 0.009j]:
    img = julia_set(lambda z, c=c: z**2 + c)
    plot_julia(img, 2.1, f"2.9 в) Множество Жюлиа $f_c(z)=z^2+c$, $c={c}$")

# г) обобщённые рациональные итерации
f1 = lambda z: (1j*z**5 - 1) / (z**5 - 1j)
f2 = lambda z: (z**2 + 1j) / (1 - z**2)

img1 = julia_set(f1)
plot_julia(img1, 2.1, "2.9 г) $f(z)=\\dfrac{iz^5-1}{z^5-i}$")

img2 = julia_set(f2)
plot_julia(img2, 2.1, "2.9 г) $f(z)=\\dfrac{z^2+i}{1-z^2}$")
```

**2.10\*** Условие: Напишите программу с использованием инструментария Jupyter Notebook, которая изображает на сфере Римана следующие кривые: a) прямую; б) окружность; в) параболу; г) гиперболу.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

def stereographic(z):
    x, y = z.real, z.imag
    d = 1 + x**2 + y**2
    return 2*x/d, 2*y/d, (x**2 + y**2 - 1)/d

def sphere_mesh():
    u = np.linspace(0, 2*np.pi, 60)
    v = np.linspace(0, np.pi, 30)
    xs = np.outer(np.cos(u), np.sin(v))
    ys = np.outer(np.sin(u), np.sin(v))
    zs = np.outer(np.ones_like(u), np.cos(v))
    return xs, ys, zs

# a) прямая y = x
t = np.linspace(-30, 30, 4000)
line = t + 1j*t

# б) окружность |z-1| = 1.5
theta = np.linspace(0, 2*np.pi, 2000)
circle = 1 + 1.5*np.exp(1j*theta)

# в) парабола y = x^2/4
tp = np.linspace(-8, 8, 4000)
parabola = tp + 1j*(tp**2/4)

# г) гипербола x*y = 1 (обе ветви)
th_pos = np.linspace(0.05, 12, 2000)
hyp1 = th_pos + 1j*(1/th_pos)
hyp2 = -th_pos - 1j*(1/th_pos)
hyperbola = np.concatenate([hyp1, hyp2])

curves = [(line, "a) прямая $y=x$"), (circle, "б) окружность $|z-1|=1.5$"),
          (parabola, "в) парабола $y=x^2/4$"), (hyperbola, "г) гипербола $xy=1$")]

xs, ys, zs = sphere_mesh()
fig = plt.figure(figsize=(11, 11))
for i, (z_curve, label) in enumerate(curves):
    ax = fig.add_subplot(2, 2, i + 1, projection='3d')
    ax.plot_surface(xs, ys, zs, color='lightblue', alpha=0.2, linewidth=0)
    X, Y, Z = stereographic(z_curve)
    ax.plot(X, Y, Z, color='crimson', lw=2)
    ax.scatter([0], [0], [1], color='black', s=20)
    ax.set_title(label, fontsize=10)
    ax.set_box_aspect([1, 1, 1])

fig.suptitle("2.10 Образы кривых на сфере Римана (стереографическая проекция)")
plt.tight_layout()
plt.show()
```

**2.11\*** Условие: Изобразите на сфере Римана логарифмическую спираль, заданную уравнением $r = ae^{b\varphi}$.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

a_param, b_param = 1.0, 0.15
phi = np.linspace(-4*np.pi, 4*np.pi, 4000)
r = a_param*np.exp(b_param*phi)
spiral = r*np.exp(1j*phi)

xs, ys, zs = sphere_mesh()
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(xs, ys, zs, color='lightblue', alpha=0.25, linewidth=0)
X, Y, Z = stereographic(spiral)
ax.plot(X, Y, Z, color='darkorange', lw=1.8)
ax.scatter([0], [0], [1], color='black', s=25)
ax.text(0, 0, 1.05, 'N')
ax.set_box_aspect([1, 1, 1])
ax.set_title(f"2.11 Логарифмическая спираль $r=ae^{{b\\varphi}}$, $a={a_param}$, $b={b_param}$\nна сфере Римана")
plt.show()
```
